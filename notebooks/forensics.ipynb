```python
import pandas as pd
import numpy as np
```

```python
timestamp = "2026-07-11_17-38-05"
data = pd.read_parquet(path=f"../target/{timestamp}_record.parquet")
```

```python
data.keys()
```

```python
max_time_s = np.max(data["time_s"])
max_time_s
```

```python
data = data[data["time_s"] > max_time_s - 25]
```

```python
cols = [c for c in data.columns if c.startswith("allied_patrol_1_")]
to_send=data[["time_s"] + cols].dropna(how="all", subset=cols)
to_send.to_csv(f"../target/{timestamp}_log.csv")
data.to_csv(f"../target/{timestamp}_complete_log.csv")
```
